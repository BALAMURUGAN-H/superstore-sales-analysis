```python
import matplotlib.pyplot as plt
import numpy as np

# =========================================================
# SUPERSTORE SALES ANALYSIS — PORTFOLIO DASHBOARD
# =========================================================

# -----------------------------
# KPI Calculations
# -----------------------------

total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
total_customers = df['Customer ID'].nunique()

profit_margin = (total_profit / total_sales) * 100


# -----------------------------
# Analysis Calculations
# -----------------------------

region_sales = (
    df.groupby('Region')['Sales']
      .sum()
      .sort_values(ascending=False)
)

region_profit = (
    df.groupby('Region')['Profit']
      .sum()
      .sort_values(ascending=False)
)

category_sales = (
    df.groupby('Category')['Sales']
      .sum()
      .sort_values(ascending=False)
)

category_profit = (
    df.groupby('Category')['Profit']
      .sum()
      .sort_values(ascending=False)
)

segment_sales = (
    df.groupby('Segment')['Sales']
      .sum()
      .sort_values(ascending=True)
)

subcategory_profit = (
    df.groupby('Sub-Category')['Profit']
      .sum()
      .sort_values()
)

loss_making = subcategory_profit[subcategory_profit < 0]

monthly_sales = (
    df.groupby(df['Order Date'].dt.to_period('M'))['Sales']
      .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)


# =========================================================
# CREATE DASHBOARD
# =========================================================

fig = plt.figure(figsize=(18, 22))

gs = fig.add_gridspec(
    5, 2,
    height_ratios=[1.1, 2.2, 2.2, 2.2, 2.0],
    hspace=0.42,
    wspace=0.25
)


# =========================================================
# MAIN TITLE
# =========================================================

fig.suptitle(
    'SUPERSTORE SALES ANALYSIS',
    fontsize=24,
    fontweight='bold',
    y=0.985
)

fig.text(
    0.5,
    0.965,
    'Sales Performance • Profitability • Customer & Regional Analysis',
    ha='center',
    fontsize=11
)


# =========================================================
# KPI CARDS
# =========================================================

kpi_ax = fig.add_subplot(gs[0, :])
kpi_ax.axis('off')

kpis = [
    ('TOTAL SALES', f'${total_sales/1_000_000:.2f}M'),
    ('TOTAL PROFIT', f'${total_profit/1_000:.2f}K'),
    ('PROFIT MARGIN', f'{profit_margin:.1f}%'),
    ('TOTAL ORDERS', f'{total_orders:,}'),
    ('TOTAL CUSTOMERS', f'{total_customers:,}')
]

positions = np.linspace(0.10, 0.90, 5)

for position, (label, value) in zip(positions, kpis):

    # Card
    kpi_ax.text(
        position,
        0.58,
        value,
        ha='center',
        va='center',
        fontsize=20,
        fontweight='bold',
        transform=kpi_ax.transAxes
    )

    kpi_ax.text(
        position,
        0.25,
        label,
        ha='center',
        va='center',
        fontsize=9,
        fontweight='bold',
        transform=kpi_ax.transAxes
    )


# =========================================================
# CHART 1 — SALES BY REGION
# =========================================================

ax1 = fig.add_subplot(gs[1, 0])

bars = ax1.bar(
    region_sales.index,
    region_sales.values
)

ax1.set_title(
    'Sales by Region',
    fontsize=14,
    fontweight='bold'
)

ax1.set_ylabel('Sales ($)')

for bar, value in zip(bars, region_sales.values):

    ax1.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'${value/1000:.0f}K',
        ha='center',
        va='bottom',
        fontsize=9
    )

ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

ax1.grid(
    axis='y',
    linestyle='--',
    alpha=0.25
)


# =========================================================
# CHART 2 — PROFIT BY REGION
# =========================================================

ax2 = fig.add_subplot(gs[1, 1])

bars = ax2.bar(
    region_profit.index,
    region_profit.values
)

ax2.set_title(
    'Profit by Region',
    fontsize=14,
    fontweight='bold'
)

ax2.set_ylabel('Profit ($)')

for bar, value in zip(bars, region_profit.values):

    ax2.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'${value/1000:.0f}K',
        ha='center',
        va='bottom',
        fontsize=9
    )

ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)

ax2.grid(
    axis='y',
    linestyle='--',
    alpha=0.25
)


# =========================================================
# CHART 3 — SALES BY CATEGORY
# =========================================================

ax3 = fig.add_subplot(gs[2, 0])

bars = ax3.bar(
    category_sales.index,
    category_sales.values
)

ax3.set_title(
    'Sales by Category',
    fontsize=14,
    fontweight='bold'
)

ax3.set_ylabel('Sales ($)')

for bar, value in zip(bars, category_sales.values):

    ax3.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'${value/1000:.0f}K',
        ha='center',
        va='bottom',
        fontsize=9
    )

ax3.spines['top'].set_visible(False)
ax3.spines['right'].set_visible(False)

ax3.grid(
    axis='y',
    linestyle='--',
    alpha=0.25
)


# =========================================================
# CHART 4 — PROFIT BY CATEGORY
# =========================================================

ax4 = fig.add_subplot(gs[2, 1])

bars = ax4.bar(
    category_profit.index,
    category_profit.values
)

ax4.set_title(
    'Profit by Category',
    fontsize=14,
    fontweight='bold'
)

ax4.set_ylabel('Profit ($)')

for bar, value in zip(bars, category_profit.values):

    ax4.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'${value/1000:.0f}K',
        ha='center',
        va='bottom',
        fontsize=9
    )

ax4.axhline(
    0,
    linewidth=1
)

ax4.spines['top'].set_visible(False)
ax4.spines['right'].set_visible(False)

ax4.grid(
    axis='y',
    linestyle='--',
    alpha=0.25
)


# =========================================================
# CHART 5 — MONTHLY SALES TREND
# =========================================================

ax5 = fig.add_subplot(gs[3, 0])

ax5.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o',
    linewidth=1.8,
    markersize=4
)

ax5.set_title(
    'Monthly Sales Trend',
    fontsize=14,
    fontweight='bold'
)

ax5.set_xlabel('Month')
ax5.set_ylabel('Sales ($)')

# Show selected labels
step = 6

ax5.set_xticks(
    range(0, len(monthly_sales), step)
)

ax5.set_xticklabels(
    monthly_sales.index[::step],
    rotation=45,
    ha='right'
)

ax5.grid(
    axis='y',
    linestyle='--',
    alpha=0.25
)

ax5.spines['top'].set_visible(False)
ax5.spines['right'].set_visible(False)


# =========================================================
# CHART 6 — SALES BY CUSTOMER SEGMENT
# =========================================================

ax6 = fig.add_subplot(gs[3, 1])

bars = ax6.barh(
    segment_sales.index,
    segment_sales.values
)

ax6.set_title(
    'Sales by Customer Segment',
    fontsize=14,
    fontweight='bold'
)

ax6.set_xlabel('Sales ($)')

for bar, value in zip(bars, segment_sales.values):

    ax6.text(
        value,
        bar.get_y() + bar.get_height() / 2,
        f'${value/1_000_000:.2f}M',
        va='center',
        fontsize=9
    )

ax6.spines['top'].set_visible(False)
ax6.spines['right'].set_visible(False)

ax6.grid(
    axis='x',
    linestyle='--',
    alpha=0.25
)


# =========================================================
# CHART 7 — LOSS-MAKING SUB-CATEGORIES
# =========================================================

ax7 = fig.add_subplot(gs[4, :])

bars = ax7.barh(
    loss_making.index,
    loss_making.values
)

ax7.set_title(
    'Loss-Making Sub-Categories',
    fontsize=14,
    fontweight='bold'
)

ax7.set_xlabel('Profit ($)')

ax7.axvline(
    0,
    linewidth=1.2
)

for bar, value in zip(bars, loss_making.values):

    ax7.text(
        value - 100,
        bar.get_y() + bar.get_height() / 2,
        f'-${abs(value)/1000:.1f}K',
        va='center',
        ha='right',
        fontsize=9
    )

ax7.spines['top'].set_visible(False)
ax7.spines['right'].set_visible(False)

ax7.grid(
    axis='x',
    linestyle='--',
    alpha=0.25
)


# =========================================================
# FOOTER
# =========================================================

fig.text(
    0.5,
    0.015,
    'Source: Superstore Dataset | Data Analysis using Python & Pandas',
    ha='center',
    fontsize=9
)


# =========================================================
# FINAL LAYOUT + SAVE
# =========================================================

plt.savefig(
    'superstore_sales_dashboard.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
